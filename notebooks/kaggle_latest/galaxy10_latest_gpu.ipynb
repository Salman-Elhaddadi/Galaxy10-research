```python
from galaxy10.model.train import *
seed_everything(0)
settings = RunSettings(device="cuda")
ds = Galaxy10Dataset(np.load(ROOT/"splits/train_idx.npy")[:32], transform=train_tf)
loader = make_loader(ds, settings, shuffle=True)
smoke = {}
for name, cls in [("a", ModelA), ("b", ModelB)]:
    model = cls().cuda()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    metrics = run_epoch(model, loader, "cuda", optimizer)
    assert np.isfinite(metrics["loss"])
    checkpoint = ROOT / f"smoke_{name}.pt"
    save_best_checkpoint(checkpoint, model, optimizer, epoch=1, validation=metrics, best_score=-1, config={"smoke_only":True})
    restored = torch.load(checkpoint, map_location="cuda", weights_only=False)
    model.load_state_dict(restored["model_state_dict"])
    smoke[name] = metrics
    print("GPU forward/backward/checkpoint passed:", name, metrics["loss"])
    del model, optimizer, restored
    torch.cuda.empty_cache()
write_json(ROOT/"smoke_results.json", smoke)
subprocess.run([sys.executable, "scripts/check_model_sanity.py"], check=True)
```

```python
# Original settings: 30 epochs, batch 32, Adam 1e-3, seeds 0/1/2.
# Each run performs clean validation and saves the best macro-F1 checkpoint.
for seed in (0, 1, 2):
    for model_name in ("a", "b"):
        subprocess.run([sys.executable, "-u", "experiments/train.py", "--model", model_name, "--seed", str(seed), "--device", "cuda"], check=True)
```
